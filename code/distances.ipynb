```python
cpu_index = faiss.read_index('../corpus/whole_proust/corpus_embeddings.index')
```

```python
print([method for method in dir(cpu_index) if not method.startswith("_")])
```

```python
help(cpu_index.reconstruct_batch)
```

```python
sentence_embeddings = cpu_index.reconstruct_batch(range(0,longueur))
print([method for method in dir(faiss.index) if not method.startswith("_")])
```

```python
k = longueur
D, I = cpu_index.search(sentence_embeddings[:], k)
```

```python
print(I[:,0])
```

```python
print(D[:,0])

closest_neighbours = D[:, 7]
min_idx = closest_neighbours.argmin()

sentence_a = min_idx
sentence_b = I[min_idx, 1]  # The nearest neighbor of sentence_a

print(phrase_list[sentence_a])
print(phrase_list[sentence_b])
```
